## To View a Single Game Run

```python
import plotly.express as px
import plotly.graph_objects as go
import json
import pandas as pd
import numpy as np
from typing import List, Dict, Any
from pprint import pprint as pp
```

```python
style = "5"
model = "deepseek-r1-0528"
reasoning = "high" # "medium", "low"
logs_paths = f'../src/optimal_explorer/strategies/combination_lock/logs/llm_calls/style{style}_{model}_{reasoning}.jsonl'

logs = []

with open(logs_paths, 'r') as f:
    for line in f:
        data = json.loads(line)
        logs.append(data)
```

```python
logs_df = pd.DataFrame(logs)
```
