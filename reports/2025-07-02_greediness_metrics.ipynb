```python
def compute_greediness(results_df):
    model_stats = {}

    for _, row in results_df.iterrows():
        model = row['model']
        history = row['history']

        # Count total predictions and correct ones
        total_predictions = 0
        correct_predictions = 0

        for entry in history:
            feedback = entry['feedback']  # List like [0, 2, 2]
            total_predictions += len(feedback)
            correct_predictions += sum(1 for f in feedback if f == 2)

        if model not in model_stats:
            model_stats[model] = {'correct': 0, 'total': 0}

        model_stats[model]['correct'] += correct_predictions
        model_stats[model]['total'] += total_predictions

    # Build DataFrame with greediness
    data = []
    for model, stats in model_stats.items():
        greediness = stats['correct'] / stats['total'] if stats['total'] > 0 else 0
        data.append({'model': model, 'greediness': greediness})

    return pd.DataFrame(data)
```

```python
greediness_df = compute_greediness(results_df)
```

```python
# Split the model column into model and style
greediness_df[['model_base', 'style']] = greediness_df['model'].str.extract(r'(.+?)\s*\(s=(\d+_default)\)')
greediness_df = greediness_df[['model_base', 'style', 'greediness']].rename(columns={'model_base': 'model'})
# remove the _default from the style column
greediness_df['style'] = greediness_df['style'].str.replace('_default', '')
```
